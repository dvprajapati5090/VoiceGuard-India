```python
import pandas as pd
import librosa
import numpy as np
import io
import pickle
import xgboost as xgb
import warnings
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split
from tqdm import tqdm
from huggingface_hub import HfFileSystem, hf_hub_download
warnings.filterwarnings('ignore')

HF_TOKEN = 'hf_YOUR_TOKEN_HERE'

def extract_features(audio_dict, sr=16000):
    try:
        y, sr = librosa.load(io.BytesIO(audio_dict['bytes']), sr=sr, duration=3.0)
        if len(y) < 1000: return None
        mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=20)
        mfcc_mean = np.mean(mfccs.T, axis=0)
        delta_mean = np.mean(librosa.feature.delta(mfccs).T, axis=0)[:10]
        centroid = np.mean(librosa.feature.spectral_centroid(y=y, sr=sr))
        zcr = np.mean(librosa.feature.zero_crossing_rate(y))
        rolloff = np.mean(librosa.feature.spectral_rolloff(y=y, sr=sr))
        rms = np.mean(librosa.feature.rms(y=y))
        return np.hstack((mfcc_mean, delta_mean, centroid, zcr, rolloff, rms))
    except: return None

fs = HfFileSystem(token=HF_TOKEN)

def get_parquet(repo_id, kw):
    try:
        files = fs.glob('datasets/' + repo_id + '/**/*.parquet')
        for f in files:
            if kw in f and 'train' in f: return f
        for f in files:
            if 'train' in f: return f
    except: pass
    return None

def pull(repo_id, label, kw, name, limit=1200):
    tag = 'HUMAN' if label == 1 else 'DEEPFAKE'
    print('--- ' + name + ' (' + tag + ') ---')
    p = get_parquet(repo_id, kw)
    if not p: print('SKIP: not found'); return [], []
    parts = p.split('/')
    try:
        lp = hf_hub_download(repo_id=parts[1]+'/'+parts[2], filename='/'.join(parts[3:]), repo_type='dataset', token=HF_TOKEN)
        df = pd.read_parquet(lp)
        Xd, yd = [], []
        for _, row in tqdm(df.iterrows(), total=min(len(df), limit)):
            if len(Xd) >= limit: break
            if 'audio' in row and row['audio'] is not None:
                feat = extract_features(row['audio'])
                if feat is not None: Xd.append(feat); yd.append(label)
        print('Got ' + str(len(Xd)) + ' samples'); return Xd, yd
    except Exception as e: print('ERROR: ' + str(e)); return [], []

X, y = [], []

for ds in [
    ('google/fleurs', 'parquet-data/hi_in/train-00000-of-00001.parquet', 'Google Fleurs', 1, 1500),
    ('SPRINGLab/IndicTTS-Hindi', 'data/train-00000-of-00010.parquet', 'IndicTTS', 1, 1000),
]:
    print('Downloading ' + ds[2] + ' (AUTHENTIC)...')
    try:
        lp = hf_hub_download(repo_id=ds[0], filename=ds[1], repo_type='dataset', token=HF_TOKEN)
        df = pd.read_parquet(lp)
        count = 0
        for _, row in tqdm(df.iterrows(), total=min(len(df), ds[4])):
            if count >= ds[4]: break
            if 'audio' in row and row['audio'] is not None:
                feat = extract_features(row['audio'])
                if feat is not None: X.append(feat); y.append(1); count += 1
        print('Got ' + str(count) + ' samples')
    except Exception as e: print('ERROR: ' + str(e))

for sfile in [
    'data/train-00000-of-00004.parquet',
    'data/train-00001-of-00004.parquet',
    'data/train-00002-of-00004.parquet',
    'data/train-00003-of-00004.parquet',
]:
    print('Downloading Sarvam ' + sfile + ' (DEEPFAKE)...')
    try:
        lp = hf_hub_download(repo_id='abhirl/hindi-tts-dataset', filename=sfile, repo_type='dataset', token=HF_TOKEN)
        df = pd.read_parquet(lp)
        count = 0
        for _, row in tqdm(df.iterrows(), total=len(df)):
            if 'audio' in row and row['audio'] is not None:
                feat = extract_features(row['audio'])
                if feat is not None: X.append(feat); y.append(0); count += 1
        print('Got ' + str(count) + ' deepfake samples from ' + sfile)
    except Exception as e: print('ERROR on ' + sfile + ': ' + str(e))

X_final = np.array(X); y_final = np.array(y)
n_auth = int(np.sum(y_final==1)); n_fake = int(np.sum(y_final==0))
print('TOTAL: ' + str(len(y_final)) + ' | Authentic: ' + str(n_auth) + ' | Deepfake: ' + str(n_fake))
if len(np.unique(y_final)) < 2: print('FATAL: Missing one class!')
else: print('Ready to train!')
```

```python
if len(np.unique(y_final)) >= 2:
    X_train, X_test, y_train, y_test = train_test_split(X_final, y_final, test_size=0.15, random_state=42, stratify=y_final)
    ratio = float(np.sum(y_train==1)) / max(float(np.sum(y_train==0)), 1.0)
    print('Class balance ratio: ' + str(round(ratio,2)))
    clf = xgb.XGBClassifier(n_estimators=500, max_depth=8, learning_rate=0.03,
                             scale_pos_weight=ratio, subsample=0.8, colsample_bytree=0.8,
                             n_jobs=-1, random_state=42, eval_metric='logloss')
    clf.fit(X_train, y_train, eval_set=[(X_test, y_test)], verbose=100)
    print('ACCURACY ON UNSEEN TEST DATA:')
    print(classification_report(y_test, clf.predict(X_test), target_names=['Deepfake','Authentic']))
    with open('voiceguard_v3_model.pkl', 'wb') as fout: pickle.dump(clf, fout)
    print('SAVED: voiceguard_v3_model.pkl')
    print('Download and rename to voiceguard_rf_model.pkl in ml_engine/models/')
```
